## Remove images that do not exist in csv file

```python
import os
import pandas as pd


def remove_nonexistent_images(csv_file: str, image_folder: str):
    """
    Removes rows from a CSV file that contain image filenames not present in the specified folder.

    Args:
        csv_file (str): The path to the CSV file.
        image_folder (str): The path to the folder containing the images.

    Returns:
        None

    Raises:
        None

    """
    # Get a list of all image files in the folder
    image_files = os.listdir(image_folder)

    # Load the CSV file
    df = pd.read_csv(csv_file)

    # Filter the dataframe to keep only the rows with image filenames that exist in the folder
    df = df[df['filename'].isin(image_files)]

    # Save the updated dataframe back to the CSV file
    df.to_csv(csv_file, index=False)

    print("Nonexistent image filenames have been removed from the CSV file.")

# Usage example
csv_file =  './selected_bahar/crops_recognizes_csv_most_freq_supp.csv'
image_folder = './selected_bahar/'
remove_nonexistent_images(csv_file, image_folder)
```

## Choose random line of text and remove them from txt file 

```python
import random

def choose_random_lines(file_path: str, num_lines: int) -> List[str]:
    """
    Chooses random lines from a text file, removes them from the file, and returns them as a list.

    Args:
        file_path (str): The path to the text file.
        num_lines (int): The number of random lines to choose.

    Returns:
        List[str]: A list of randomly chosen lines from the file.

    Raises:
        FileNotFoundError: If the specified file does not exist.

    """    
    
    lines = []
    chosen_lines = []

    # Read the text file and store each line
    with open(file_path, 'r') as file:
        lines = file.readlines()

    # Choose random lines
    random_lines = random.sample(lines, num_lines)

    # Store the randomly chosen lines in a variable
    chosen_lines = [line.strip() for line in random_lines]

    # Remove the selected lines from the file
    with open(file_path, 'w') as file:
        for line in lines:
            if line not in random_lines:
                file.write(line)

    return chosen_lines

# Usage example
txt_file = './fa_text.txt'
num_selected_lines = 100
random_lines = choose_random_lines(txt_file , num_selected_lines)
print(random_lines)  # Print the randomly chosen lines stored in the variable
```

## Transfer images that exist in csv file into a new folder

```python
import csv
import shutil
import os

def transfer_images(csv_file : str, source_folder : str, destination_folder : str) -> None:
     """
    Transfers images listed in a CSV file from a source folder to a destination folder.

    Args:
        csv_file (str): The path to the CSV file containing the list of image names.
        source_folder (str): The path to the folder where the source images are located.
        destination_folder (str): The path to the folder where the images will be transferred.

    Returns:
        None

    Raises:
        FileNotFoundError: If the CSV file or the source folder does not exist.

    """
        
    # Create the destination folder if it doesn't exist
    os.makedirs(destination_folder, exist_ok=True)

    with open(csv_file, 'r') as file:
        reader = csv.reader(file)
        next(reader)  # Skip the header row if present

        for row in reader:
            image_name = row[0]  # Assuming the image name is in the first column of the CSV file

            # Construct the source and destination paths
            source_path = os.path.join(source_folder, image_name)
            destination_path = os.path.join(destination_folder, image_name)

            # Check if the source file exists
            if os.path.exists(source_path):
                # Copy or move the image file to the destination folder
                shutil.copy(source_path, destination_path)  # Change to shutil.move if you want to move instead of copy

# Usage example
csv_file = 'Dataset_split42000/label_noisy.csv'
source_folder = './Data_42000'
destination_folder = './Dataset_split42000/Data_noisy_34000'

transfer_images(csv_file, source_folder, destination_folder)
```

## Generate random number

```python
from faker import Faker
from typing import List

def generate_fa_random_number(num_gen: int) -> List[str]:
    """
    Generates fake Persian random numbers.

    Args:
        num_gen (int): The number of fake Persian numbers to generate.

    Returns:
        List[str]: A list of generated fake Persian numbers.

    Raises:
        None

    """

    fake = Faker()
    nums = []

    my_word_list = ["۰", "۱", "۲", "۳", "۴", "۵", "۶", "۷", "۸", "۹"]
    for i in range(num_gen):
        random_num = fake.sentence(ext_word_list=my_word_list)
        
        # Avoid numbers starting with Persion or Arabic or English zero 
        if (random_num[0] != "۰") and (random_num[0] != "٠") and (random_num[0] != "0"):
            random_num = random_num.removesuffix('.')
            
            # Remove space
            random_num = random_num.replace(" ", "")
            
            # Append with three zeros 
            # random_num = random_num + '۰۰۰' 
            nums.append(random_num)
    return nums

numbers = generate_fa_random_number(num_gen=100)  
print(numbers)

# Saves numbers to a file
with open('/content/drive/MyDrive/test.txt', 'w')as f:
    for num in numbers:
        f.write(num+'\n')
```

```python

import os
import more_itertools
from faker import Faker
from typing import List

fake = Faker()

def generate_fake_sentence(word_list: List[str], num_sentence: int) -> List[str]:
    """
    Generates fake sentences using words from the given word_list.

    Args:
        word_list (List[str]): A list of words to be used for generating sentences.
        num_sentence (int): The number of sentences to generate.

    Returns:
        List[str]: A list of generated fake sentences.

    Note:
        The function uses the Faker library to generate sentences. It removes the period (.) from
        the end of each sentence and only includes sentences with a length less than 49 characters.
    """
    sentences = []
    for _ in range(num_sentence):
        generated_sen = fake.sentence(ext_word_list=word_list)
        generated_sen = generated_sen.removesuffix('.')  # Drop the dot from the sentence ending
        if len(generated_sen) < 49:
            sentences.append(generated_sen)
    return sentences

def chunk_data(word_list: List[str], num_sentence: int, chunk: int, dir_name: str,
               file_format: str = 'chunk_{:03d}.txt', overwrite: bool = False) -> None:
    """
    Generates fake sentences using words from the given word_list and saves them into multiple files.

    Args:
        word_list (List[str]): A list of words to be used for generating sentences.
        num_sentence (int): The total number of sentences to generate.
        chunk (int): The number of sentences to include in each chunk or file.
        dir_name (str): The directory name where the generated files will be saved.
        file_format (str, optional): The filename format for the generated files. Defaults to 'chunk_{:03d}.txt'.
        overwrite (bool, optional): Whether to overwrite existing files. Defaults to False.

    Returns:
        None

    Note:
        The function calls generate_fake_sentence() to generate fake sentences and then splits them
        into chunks using the more_itertools.chunked() function. Each chunk is saved as a separate file
        in the specified directory with a filename based on the file_format and chunk number.
    """
    os.makedirs(dir_name, exist_ok=True)

    sentences = generate_fake_sentence(word_list, num_sentence)
    num_chunks = (num_sentence + chunk - 1) // chunk

    for i, chunked_sentences in enumerate(more_itertools.chunked(sentences, chunk)):
        file_path = os.path.join(dir_name, file_format.format(i + 1))
        if not overwrite and os.path.exists(file_path):
            continue

        with open(file_path, 'a') as f:
            for line in chunked_sentences:
                f.write(line + '\n')

        print(f"Chunk {i + 1}/{num_chunks} saved to {file_path}")

# Determine which items should be used to generate a sentence.
my_word_list = ['مر', 'م', 'ع', '۴', 'عربی', 'مهرام', '2', 'z',
                 'g', 'و', 'ع', 'e', 'ء','گ', 'ی', 'گیره', 'قهوه ای'
                 , 'zero', '22', '202', '0020', '(گربه)', '\۰', '0|']

chunk_data(word_list=my_word_list, num_sentence=10, chunk=5, dir_name='chunk')
```

## Extract unique words and digits in a txt file

```python

import re

def extract_unique_words_and_digits(file_path: str) -> tuple:
    """
    Extracts unique words and digits from a file.

    Args:
        file_path (str): The path to the file to extract words and digits from.

    Returns:
        tuple: A tuple containing two sets:
            - The first set contains unique words extracted from the file.
            - The second set contains unique digits extracted from the file.

    Note:
        - The function reads the contents of the file specified by the file_path.
        - It uses regular expressions to extract unique words and digits from the file content.
        - Unique words are extracted using the pattern '[^\W\d_]+', which matches any sequence of letters.
        - Unique digits are extracted using the pattern '\d+', which matches any sequence of digits.
        - The function returns a tuple containing two sets: unique_words and unique_digits.
    """
    
    
    unique_words = set()
    unique_digits = set()

    with open(file_path, 'r', encoding='utf-8') as file:
        content = file.read()

        # Extract unique words
        words = re.findall(r'[^\W\d_]+', content, flags=re.UNICODE)
        unique_words = set(words)

        # Extract unique digits
        digits = re.findall(r'\d+', content)
        unique_digits = set(digits)

    return unique_words, unique_digits

# Example usage
file_path = '/content/drive/MyDrive/tmp.txt'
unique_words, unique_digits = extract_unique_words_and_digits(file_path)

# Print the unique words
print("Unique words:")
for word in unique_words:
    print(word)

# Print the unique digits
print("Unique digits:")
for digit in unique_digits:
    print(digit)
```
